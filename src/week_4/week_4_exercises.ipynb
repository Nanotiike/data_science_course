## Text generation example

This is a brief example of how to run text generation with a causal language model and `pipeline`.

Install [transformers](https://huggingface.co/docs/transformers/index) python package. This will be used to load the model and tokenizer and to run generation.

```python
!pip install --quiet transformers
```

Import the `AutoTokenizer`, `AutoModelForCausalLM`, and `pipeline` classes. The first two support loading tokenizers and generative models from the [Hugging Face repository](https://huggingface.co/models), and the last wraps a tokenizer and a model for convenience.

```python
from transformers import AutoTokenizer, AutoModelForCausalLM, pipeline
```

Load a generative model and its tokenizer. You can substitute any other generative model name here (e.g. [other TurkuNLP GPT-3 models](https://huggingface.co/models?sort=downloads&search=turkunlp%2Fgpt3)), but note that Colab may have issues running larger models.

```python
MODEL_NAME = 'gpt2-large'

tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)
model = AutoModelForCausalLM.from_pretrained(MODEL_NAME).to('cuda')
```

Instantiate a text generation pipeline using the tokenizer and model.

```python
pipe = pipeline(
    'text-generation',
    model=model,
    tokenizer=tokenizer,
    device=model.device
)
```

We can now call the pipeline with a text prompt; it will take care of tokenizing, encoding, generation, and decoding:

```python
output = pipe('Terve, miten menee?', max_new_tokens=25)

print(output)
```

Just print the text

```python
print(output[0]['generated_text'])
```

We can also call the pipeline with any arguments that the model `generate` function supports. For details on text generation using `transformers`, see e.g. [this tutorial](https://huggingface.co/blog/how-to-generate).

Example with sampling and a high `temperature` parameter to generate more chaotic output:

```python
output = pipe(
    'Terve, miten menee?',
    do_sample=True,
    temperature=10.0,
    max_new_tokens=25
)

print(output[0]['generated_text'])
```

```python
#Binary sentiment classification. zero-shot, one-shot, and two-shot

# zero-shot

output_1 = pipe(
    'Do the following text express a positive or negative sentiment? Text: I love hanging out with my friends. Last night we went Surfing and it was so much fun. Answer:',
    do_sample=True,
    temperature=1.0,
    max_new_tokens=25
)

print(output_1[0]['generated_text'])

output_2 = pipe(
    'Do the following text express a positive or negative sentiment? Text: The capitalist machine will take our souls, eat them up, and spit out the remains. The rich will take our lives and then tell us to be thankful for them. Answer:',
    do_sample=True,
    temperature=1.0,
    max_new_tokens=25
)

print(output_2[0]['generated_text'])
```

```python
# one-shot

output_1 = pipe(
    """Do the following texts express a positive or negative sentiment?
    Text: Cabbage is awful and I hate eating it. My parents should stop putting it in food so much.
    Answer: Text expresses negative sentiment.
    Text: The capitalist machine will take our souls, eat them up, and spit out the remains. The rich will take our lives and then tell us to be thankful for them.
    Answer:""",
    do_sample=True,
    temperature=1.0,
    max_new_tokens=25
)

print(output_1[0]['generated_text'])

output_2 = pipe(
    """Do the following texts express a positive or negative sentiment?
    Text: Cabbage is awful and I hate eating it. My parents should stop putting it in food so much.
    Answer: Text expresses negative sentiment.
    Text: I love hanging out with my friends. Last night we went Surfing and it was so much fun.
    Answer:""",
    do_sample=True,
    temperature=1.0,
    max_new_tokens=25
)

print(output_2[0]['generated_text'])
```

```python
# two-shot

output_1 = pipe(
    """Do the following texts express a positive or negative sentiment?
    Text: Cabbage is awful and I hate eating it. My parents should stop putting it in food so much.
    Answer: Text expresses negative sentiment.
    Text: Surfing is my favorite past time. I love going to beaches with my surfboard and catching waves for an entire day.
    Answer: Text expresses positive sentiment.
    Text: The capitalist machine will take our souls, eat them up, and spit out the remains. The rich will take our lives and then tell us to be thankful for them.
    Answers:""",
    do_sample=True,
    temperature=1.0,
    max_new_tokens=25
)

print(output_1[0]['generated_text'])

output_2 = pipe(
    """Do the following texts express a positive or negative sentiment?
    Text: Cabbage is awful and I hate eating it. My parents should stop putting it in food so much.
    Answer: Text expresses negative sentiment.
    Text: Surfing is my favorite past time. I love going to beaches with my surfboard and catching waves for an entire day.
    Answer: Text expresses positive sentiment.
    Text: I love hanging out with my friends. Last night we went Surfing and it was so much fun.
    Answers:""",
    do_sample=True,
    temperature=1.0,
    max_new_tokens=25
)

print(output_2[0]['generated_text'])
```

Perhaps a bit strangly the one-shot prompt was the best, even better than two-shot. They tend to have the tendency to add more text than is necessary, adding on after the answer. Somewhat does its job, but would need a lot of refinement. The zero-shot doesn't work at all, the model not really seeming to know what to do. Just continuing of what text there is.

```python
# Person name recognition

# zero-shot

output_1 = pipe(
    """List the person names occurring in the following texts.
    Text: Matias has been working on that paper for so long now. I think his text will be a lot better than William's, who wrote it all in one night.
    Answer:""",
    do_sample=True,
    temperature=1.0,
    max_new_tokens=25
)

print(output_1[0]['generated_text'])

output_2 = pipe(
    """List the person names occurring in the following texts.
    Text: Are Sarah and Sofia coming to the party? They promised they would. Maria is coming and I'd hate it to be just the two of us.
    Answer:""",
    do_sample=True,
    temperature=1.0,
    max_new_tokens=25
)

print(output_2[0]['generated_text'])
```

```python
# one-shot

output_1 = pipe(
    """List the person names occurring in the following texts.
    Text: What do you think of Tobias and Rick? Can they keep up with the rest of the crew? Or do they need more training?
    Answer: Tobias, Rick
    Text: Matias has been working on that paper for so long now. I think his text will be a lot better than William's, who wrote it all in one night.
    Answer:""",
    do_sample=True,
    temperature=1.0,
    max_new_tokens=25
)

print(output_1[0]['generated_text'])

output_2 = pipe(
    """List the person names occurring in the following texts.
    Text: What do you think of Tobias and Rick? Can they keep up with the rest of the crew? Or do they need more training?
    Answer: Tobias, Rick
    Text: Are Sarah and Sofia coming to the party? They promised they would. Maria is coming and I'd hate it to be just the two of us.
    Answer:""",
    do_sample=True,
    temperature=1.0,
    max_new_tokens=25
)

print(output_2[0]['generated_text'])
```

```python
# two-shot

output_1 = pipe(
    """List the person names occurring in the following texts.
    Text: What do you think of Tobias and Rick? Can they keep up with the rest of the crew? Or do they need more training?
    Answer: Tobias, Rick
    Text: Welcome to Penny and Benny's fantastical tour! Here you'll see all that is amde by the great Philip.
    Answer: Penny, Benny, Philip
    Text: Matias has been working on that paper for so long now. I think his text will be a lot better than William's, who wrote it all in one night.
    Answer:""",
    do_sample=True,
    temperature=1.0,
    max_new_tokens=25
)

print(output_1[0]['generated_text'])

output_2 = pipe(
    """List the person names occurring in the following texts.
    Text: What do you think of Tobias and Rick? Can they keep up with the rest of the crew? Or do they need more training?
    Answer: Tobias, Rick
    Text: Welcome to Penny and Benny's fantastical tour! Here you'll see all that is amde by the great Philip.
    Answer: Penny, Benny, Philip
    Text: Are Sarah and Sofia coming to the party? They promised they would. Maria is coming and I'd hate it to be just the two of us.
    Answer:""",
    do_sample=True,
    temperature=1.0,
    max_new_tokens=25
)

print(output_2[0]['generated_text'])
```

Can do this to an extent with some examples, but does still miss names. And it still has the habit of extending the text further than necessary, adding on the patern.

```python
# Two-digit addition

# zero-shot

output_1 = pipe(
    """This is a first grade math exam.
    What is 10+15?
    Answer:""",
    do_sample=True,
    temperature=1.0,
    max_new_tokens=25
)

print(output_1[0]['generated_text'])

output_2 = pipe(
    """This is a first grade math exam.
    What is 3+9?
    Answer:""",
    do_sample=True,
    temperature=1.0,
    max_new_tokens=25
)

print(output_2[0]['generated_text'])
```

```python
# one-shot

output_1 = pipe(
    """This is a first grade math exam.
    What is 5+14?
    Answer: 19
    What is 10+15?
    Answer:""",
    do_sample=True,
    temperature=1.0,
    max_new_tokens=25
)

print(output_1[0]['generated_text'])

output_2 = pipe(
    """This is a first grade math exam.
    What is 5+14?
    Answer: 19
    What is 3+9?
    Answer:""",
    do_sample=True,
    temperature=1.0,
    max_new_tokens=25
)

print(output_2[0]['generated_text'])
```

```python
# two-shot

output_1 = pipe(
    """This is a first grade math exam.
    What is 5+14?
    Answer: 19
    What is 21+7?
    Answer: 28
    What is 10+15?
    Answer:""",
    do_sample=True,
    temperature=1.0,
    max_new_tokens=25
)

print(output_1[0]['generated_text'])

output_2 = pipe(
    """This is a first grade math exam.
    What is 5+14?
    Answer: 19
    What is 21+7?
    Answer: 28
    What is 3+9?
    Answer:""",
    do_sample=True,
    temperature=1.0,
    max_new_tokens=25
)

print(output_2[0]['generated_text'])
```

The model can't really do math properly. It sees the pattern but doesn't understand the contents and can't do the addition correctly. And it continues as before.
